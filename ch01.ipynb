```
from google.colab.patches import cv2_imshow
import cv2
import os

image_path = '/content/gdrive/MyDrive/Image_processing/lena.jpg'

# 파일 존재 여부 확인 후 불러오기
if os.path.exists(image_path):
    img = cv2.imread(image_path)

    # 영상 화면에 표시하기
    cv2_imshow(img)

    # 영상을 구글 드라이브 내 동일한 폴더에 새로운 이름으로 저장하기
    output_path = '/content/gdrive/MyDrive/Image_processing/lena_copy.jpg'
    cv2.imwrite(output_path, img)
    print(f"성공적으로 이미지를 불러오고 복사본을 저장했습니다!\n저장된 위치: {output_path}")
else:
    print(f"오류: 해당 경로에 파일이 존재하지 않습니다.\n확인한 경로: {image_path}")
```
